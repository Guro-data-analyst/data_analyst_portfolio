#### Стратегия обработки:

1. **Year of Release** - удалить строки с пропусками, так как год важен для фильтрации

2. **Critic Score и User Score** - заменить пропуски на -1 как индикатор отсутствия оценки

- Обработайте пропущенные значения. Для каждого случая вы можете выбрать оптимальный, на ваш взгляд, вариант: заменить на определённое значение, оставить как есть или удалить.
- Если вы решите заменить пропуски на значение-индикатор, то убедитесь, что предложенное значение не может быть использовано в данных.
- Если вы нашли пропуски в данных с количеством проданных копий игры в том или ином регионе, их можно заменить на среднее значение в зависимости от названия платформы и года выхода игры.

```python
# Обработка пропусков
print("Обработка пропусков...")
initial_rows = len(df)

# 1. Удаляем строки с пропусками в name и genre
df = df.dropna(subset=['name', 'genre'])
print(f"Удалено строк с пропусками в name/genre: {initial_rows - len(df)}")

# 2. Удаляем строки с пропусками в year_of_release
initial_rows = len(df)
df = df.dropna(subset=['year_of_release'])
print(f"Удалено строк с пропусками в year_of_release: {initial_rows - len(df)}")

# 3. Преобразуем year_of_release в int
df['year_of_release'] = df['year_of_release'].astype(int)

# 4. Заменяем пропуски в оценках на -1
df['critic_score'] = df['critic_score'].fillna(-1)
df['user_score'] = df['user_score'].fillna(-1)

# 5. Заменяем пропуски в рейтинге на 'UNKNOWN'
df['rating'] = df['rating'].fillna('UNKNOWN')

# 6. Заменяем пропуски в продажах на 0
df['eu_sales'] = df['eu_sales'].fillna(0)
df['jp_sales'] = df['jp_sales'].fillna(0)

print("\nПроверка пропусков после обработки:")
print(df.isnull().sum())
```

## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
# 1. Категоризация по оценкам пользователей
print("Категоризация по оценкам пользователей...")

def categorize_user_score(score):
    if score == -1:
        return 'нет оценки'
    elif 0 <= score < 3:
        return 'низкая оценка'
    elif 3 <= score < 8:
        return 'средняя оценка'
    elif 8 <= score <= 10:
        return 'высокая оценка'
    else:
        return 'некорректная оценка'

df_actual['user_score_category'] = df_actual['user_score'].apply(categorize_user_score)

print("Распределение по категориям пользовательских оценок:")
user_dist = df_actual['user_score_category'].value_counts()
display(user_dist)
```

- Разделите все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
# 2. Категоризация по оценкам критиков
print("\nКатегоризация по оценкам критиков...")

def categorize_critic_score(score):
    if score == -1:
        return 'нет оценки'
    elif 0 <= score < 30:
        return 'низкая оценка'
    elif 30 <= score < 80:
        return 'средняя оценка'
    elif 80 <= score <= 100:
        return 'высокая оценка'
    else:
        return 'некорректная оценка'

df_actual['critic_score_category'] = df_actual['critic_score'].apply(categorize_critic_score)

print("Распределение по категориям оценок критиков:")
critic_dist = df_actual['critic_score_category'].value_counts()
display(critic_dist)
```

- После категоризации данных проверьте результат: сгруппируйте данные по выделенным категориям и посчитайте количество игр в каждой категории.

```python
# Проверка результатов категоризации
print("Сводная информация по категоризации:")

summary_df = pd.DataFrame({
    'Пользовательские оценки': user_dist,
    'Оценки критиков': critic_dist
})

display(summary_df)

# Анализ игр с оценками (исключая "нет оценки")
print("\nАнализ игр, имеющих оценки:")
print(f"Игры с пользовательскими оценками: {len(df_actual[df_actual['user_score'] != -1])}")
print(f"Игры с оценками критиков: {len(df_actual[df_actual['critic_score'] != -1])}")
print(f"Игры с обеими оценками: {len(df_actual[(df_actual['user_score'] != -1) & (df_actual['critic_score'] != -1)])}")
```

- Выделите топ-7 платформ по количеству игр, выпущенных за весь актуальный период.

```python
# 3. Топ-7 платформ по количеству выпущенных игр
print("\nОпределение топ-7 платформ по количеству игр...")

# Считаем количество игр для каждой платформы
platform_counts = df_actual['platform'].value_counts().head(7)

print("Топ-7 платформ (2000-2013 гг.):")
for i, (platform, count) in enumerate(platform_counts.items(), 1):
    print(f"{i}. {platform}: {count} игр")

# Сохраняем список топ-платформ для дальнейшего использования
top_platforms = platform_counts.index.tolist()

print(f"\nТоп-платформы охватывают {platform_counts.sum()} игр из {len(df_actual)}")
print(f"Доля топ-платформ: {(platform_counts.sum() / len(df_actual) * 100):.1f}%")

# Визуализация в виде датафрейма
top_platforms_df = platform_counts.reset_index()
top_platforms_df.columns = ['platform', 'game_count']
display(top_platforms_df)
```
